## 2. Fundamento teórico

### 2.1. Resolución lógica

La resolución es una regla de inferencia que permite
obtener nuevas cláusulas a partir de dos cláusulas que
contienen literales complementarios.

Por ejemplo:

$$A \lor B$$

$$\neg B \lor C$$

Al eliminar los literales complementarios B y ¬B,
se obtiene:

$$A \lor C$$

La nueva cláusula recibe el nombre de resolvente.

Este procedimiento permite demostrar consecuencias
lógicas sin enumerar explícitamente todos los modelos
posibles de una base de conocimiento.



### 2.2. Conceptos fundamentales

| Concepto | Definición | Ejemplo |
|---|---|---|
| Átomo | Proposición elemental. | P |
| Literal positivo | Proposición sin negación. | P |
| Literal negativo | Proposición negada. | ¬P |
| Literales complementarios | Literales opuestos entre sí. | P y ¬P |
| Cláusula | Disyunción de uno o más literales, o cláusula vacía. | P ∨ Q |
| Cláusula unitaria | Cláusula con un único literal. | P |
| Cláusula vacía | Cláusula que representa una contradicción. | □ |
| Resolvente | Cláusula obtenida mediante resolución. | A ∨ C |
| Tautología | Expresión verdadera bajo cualquier interpretación. | P ∨ ¬P |



### 2.3. Forma normal conjuntiva (FNC)

La forma normal conjuntiva es una representación lógica
constituida por una conjunción de cláusulas, donde cada
cláusula es una disyunción de literales.

Ejemplo:

$$(A \lor B)\land(\neg C \lor D)\land E$$

Para convertir una expresión lógica a FNC se realizan
los siguientes pasos:

1. Eliminar los bicondicionales.
2. Eliminar las implicaciones.
3. Aplicar las leyes de De Morgan y eliminar dobles negaciones.
4. Distribuir la disyunción sobre la conjunción.

#### Equivalencias lógicas

**Bicondicional:**

$$A\Leftrightarrow B\equiv(A\Rightarrow B)\land(B\Rightarrow A)$$

**Implicación:**

$$A\Rightarrow B\equiv\neg A\lor B$$

**Leyes de De Morgan:**

$$\neg(A\land B)\equiv\neg A\lor\neg B$$

$$\neg(A\lor B)\equiv\neg A\land\neg B$$

**Doble negación:**

$$\neg(\neg A)\equiv A$$

**Ley distributiva:**

$$A\lor(B\land C)\equiv(A\lor B)\land(A\lor C)$$



### 2.4. Demostración por refutación

La resolución por refutación consiste en demostrar una
proposición suponiendo inicialmente que es falsa.

Para demostrar:

$$KB\models\alpha$$

Se incorpora la negación de la meta a la base de conocimiento:

$$KB\land\neg\alpha$$

Posteriormente, se aplica la resolución lógica.

Si se obtiene la cláusula vacía, significa que existe
una contradicción y que la proposición original es
consecuencia lógica de la base de conocimiento.

Si el procedimiento termina sin generar nuevas cláusulas
y sin encontrar la cláusula vacía, la proposición no es
consecuencia lógica de la KB.



## 3. Preguntas previas

### 1. ¿Por qué se niega la meta en lugar de intentar derivarla directamente?

Porque la resolución por refutación busca demostrar que
la negación de la meta es incompatible con los hechos
y reglas de la base de conocimiento.

Si agregar la meta negada produce una contradicción,
se demuestra que la meta original es una consecuencia
lógica de la KB.

### 2. ¿Qué significa obtener la cláusula vacía?

Significa que se ha encontrado una contradicción entre
las cláusulas.

La cláusula vacía se representa mediante el símbolo □
y es falsa bajo cualquier interpretación.

Su aparición demuestra que la base de conocimiento
junto con la meta negada es insatisfacible.

### 3. ¿Cuál es la FNC de ¬(A ∧ B)?

Aplicando una de las leyes de De Morgan:

$$\neg(A\land B)$$

$$\equiv\neg A\lor\neg B$$

Por lo tanto, la forma normal conjuntiva es:

$$\boxed{\neg A\lor\neg B}$$

Esta expresión constituye una única cláusula formada
por dos literales negativos.

### 4.1. Conversión a forma normal conjuntiva (FNC)

Se convertirán tres expresiones proposicionales a FNC
mediante la aplicación de equivalencias lógicas.

El objetivo es obtener cláusulas que puedan ser utilizadas
posteriormente por el algoritmo de resolución.



#### Ejercicio (a)

Convertir a FNC:

$$B_{11}\Leftrightarrow(P_{12}\lor P_{21})$$

**Paso 1. Eliminar el bicondicional.**

$$(B_{11}\Rightarrow(P_{12}\lor P_{21}))\land((P_{12}\lor P_{21})\Rightarrow B_{11})$$

**Paso 2. Eliminar las implicaciones.**

$$(\neg B_{11}\lor P_{12}\lor P_{21})\land(\neg(P_{12}\lor P_{21})\lor B_{11})$$

**Paso 3. Aplicar la ley de De Morgan.**

$$(\neg B_{11}\lor P_{12}\lor P_{21})\land((\neg P_{12}\land\neg P_{21})\lor B_{11})$$

**Paso 4. Aplicar la ley distributiva.**

$$(\neg B_{11}\lor P_{12}\lor P_{21})\land(\neg P_{12}\lor B_{11})\land(\neg P_{21}\lor B_{11})$$

**Resultado final:**

$$\boxed{(\neg B_{11}\lor P_{12}\lor P_{21})\land(\neg P_{12}\lor B_{11})\land(\neg P_{21}\lor B_{11})}$$

Se obtienen tres cláusulas:

- C1: ¬B11 ∨ P12 ∨ P21
- C2: ¬P12 ∨ B11
- C3: ¬P21 ∨ B11



#### Ejercicio (b)

Convertir a FNC:

$$\neg(P\Rightarrow(Q\land R))$$

**Paso 1. Eliminar la implicación.**

$$\neg(\neg P\lor(Q\land R))$$

**Paso 2. Aplicar la ley de De Morgan.**

$$\neg(\neg P)\land\neg(Q\land R)$$

**Paso 3. Eliminar la doble negación.**

$$P\land\neg(Q\land R)$$

**Paso 4. Aplicar nuevamente la ley de De Morgan.**

$$P\land(\neg Q\lor\neg R)$$

**Resultado final:**

$$\boxed{P\land(\neg Q\lor\neg R)}$$

Se obtienen dos cláusulas:

- C1: P
- C2: ¬Q ∨ ¬R



#### Ejercicio (c)

Convertir a FNC:

$$(A\land B)\Rightarrow C$$

**Paso 1. Eliminar la implicación.**

$$\neg(A\land B)\lor C$$

**Paso 2. Aplicar la ley de De Morgan.**

$$(\neg A\lor\neg B)\lor C$$

**Paso 3. Aplicar la propiedad asociativa.**

$$\neg A\lor\neg B\lor C$$

**Resultado final:**

$$\boxed{\neg A\lor\neg B\lor C}$$

Se obtiene una única cláusula:

- C1: ¬A ∨ ¬B ∨ C

| Ejercicio | Expresión original | Forma normal conjuntiva | Cláusulas |

| b | $\neg(P\Rightarrow(Q\land R))$ | $P\land(\neg Q\lor\neg R)$ | 2 |
| c | $(A\land B)\Rightarrow C$ | $\neg A\lor\neg B\lor C$ | 1 |

Las expresiones obtenidas conservan la equivalencia

### 4.2. Demostraciones por resolución

Para demostrar que una proposición es consecuencia
lógica de una base de conocimiento, se agrega la
negación de la meta y se aplica la regla de resolución
hasta obtener una contradicción.

Cada cláusula se identificará mediante un número
para registrar el origen de los resolventes.



#### Ejercicio A

**Base de conocimiento:**

$$KB=\{\neg B_{11}, B_{11}\Leftrightarrow(P_{12}\lor P_{21})\}$$

**Meta:**

$$KB\models\neg P_{12}$$

**Paso 1. Convertir la KB a FNC.**

| Cláusula | Expresión | Origen |
|---|---|---|
| C1 | $\neg B_{11}$ | Hecho de la KB |
| C2 | $\neg B_{11}\lor P_{12}\lor P_{21}$ | Bicondicional |
| C3 | $\neg P_{12}\lor B_{11}$ | Bicondicional |
| C4 | $\neg P_{21}\lor B_{11}$ | Bicondicional |

**Paso 2. Negar la meta.**

$$\neg(\neg P_{12})\equiv P_{12}$$

Se agrega la cláusula:

$$C5=P_{12}$$

**Paso 3. Aplicar resolución.**

Primera resolución: C1 y C3.

$$\frac{\neg B_{11}\qquad\neg P_{12}\lor B_{11}}{\neg P_{12}}$$

$$C6=\neg P_{12}$$

Segunda resolución: C5 y C6.

$$\frac{P_{12}\qquad\neg P_{12}}{\square}$$

$$C7=\square$$

**Paso 4. Registrar los resolventes.**

| Resolvente | Cláusulas utilizadas | Literal eliminado | Resultado |
|---|---|---|---|
| C6 | C1 y C3 | B11 | $\neg P_{12}$ |
| C7 | C5 y C6 | P12 | $\square$ |

**Conclusión:**

Se obtuvo la cláusula vacía después de dos resoluciones.

$$\boxed{KB\models\neg P_{12}}$$

Por lo tanto, la casilla [1,2] está libre de pozos.



#### Ejercicio B

**Base de conocimiento:**

$$KB=\{P\Rightarrow Q,\ Q\Rightarrow R,\ R\Rightarrow S,\ P\}$$

**Meta:**

$$KB\models S$$

**Paso 1. Convertir las implicaciones a FNC.**

$$P\Rightarrow Q\equiv\neg P\lor Q$$

$$Q\Rightarrow R\equiv\neg Q\lor R$$

$$R\Rightarrow S\equiv\neg R\lor S$$

| Cláusula | Expresión |
|---|---|
| C1 | $\neg P\lor Q$ |
| C2 | $\neg Q\lor R$ |
| C3 | $\neg R\lor S$ |
| C4 | $P$ |

**Paso 2. Negar la meta.**

$$C5=\neg S$$

**Paso 3. Aplicar resolución.**

Primera resolución: C1 y C4.

$$\frac{\neg P\lor Q\qquad P}{Q}$$

$$C6=Q$$

Segunda resolución: C2 y C6.

$$\frac{\neg Q\lor R\qquad Q}{R}$$

$$C7=R$$

Tercera resolución: C3 y C7.

$$\frac{\neg R\lor S\qquad R}{S}$$

$$C8=S$$

Cuarta resolución: C5 y C8.

$$\frac{\neg S\qquad S}{\square}$$

$$C9=\square$$

**Paso 4. Registrar los resolventes.**

| Resolvente | Cláusulas utilizadas | Literal eliminado | Resultado |
|---|---|---|---|
| C6 | C1 y C4 | P | Q |
| C7 | C2 y C6 | Q | R |
| C8 | C3 y C7 | R | S |
| C9 | C5 y C8 | S | $\square$ |

**Conclusión:**

Se obtuvo la cláusula vacía después de cuatro resoluciones.

$$\boxed{KB\models S}$$

Por lo tanto, S es consecuencia lógica de la base
de conocimiento.

### 4.3. Representación de cláusulas en Python

Para implementar la resolución proposicional se utilizarán
conjuntos de Python.

Los literales positivos se representarán mediante cadenas
de texto, por ejemplo, "P".

Los literales negativos se representarán utilizando el
símbolo "~", por ejemplo, "~P".

Cada cláusula se representará mediante un frozenset,
es decir, un conjunto inmutable.

| Expresión lógica | Representación en Python |
|---|---|
| $P$ | frozenset({"P"}) |
| $\neg P$ | frozenset({"~P"}) |
| $P\lor Q$ | frozenset({"P", "Q"}) |
| $\neg P\lor Q$ | frozenset({"~P", "Q"}) |
| $\square$ | frozenset() |

El uso de frozenset permite almacenar cláusulas dentro
de otros conjuntos, evitar duplicados y comparar
resolventes.


```python

# Conjunto mutable
conjunto = {"P", "Q"}

# Conjunto inmutable
clausula = frozenset({"P", "Q"})

print("Conjunto:", conjunto)
print("Tipo:", type(conjunto))

print("\nCláusula:", clausula)
print("Tipo:", type(clausula))

# Base de conocimiento como conjunto de cláusulas
kb_ejemplo = {
    frozenset({"P", "Q"}),
    frozenset({"~P", "R"}),
    frozenset({"P", "Q"})
}

print("\nBase de conocimiento:", kb_ejemplo)
print("Cantidad de cláusulas:", len(kb_ejemplo))
```


### 4.4. Implementación de la función neg()

La función neg() recibe un literal y devuelve su
complementario.

Si el literal es positivo, agrega el símbolo "~".

Si el literal es negativo, elimina el símbolo "~".

Por ejemplo:

$$neg(P)=\neg P$$

$$neg(\neg P)=P$$

Esta función permite identificar los literales
complementarios necesarios para aplicar la resolución.


```python

def neg(l):
    """Devuelve el literal complementario."""

    return l[1:] if l.startswith("~") else "~" + l


# Pruebas individuales
literales = ["P", "~P", "Q", "~Q", "B11", "~B11"]

print("LITERAL     COMPLEMENTARIO")
print("-" * 30)

for literal in literales:
    print(f"{literal:12}{neg(literal)}")
```


### 4.5. Implementación de la función resolver()

La función resolver() recibe dos cláusulas y genera
sus posibles resolventes.

Para cada literal de la primera cláusula se busca
su complementario en la segunda.

Si se encuentra un par de literales complementarios,
se eliminan ambos y se realiza la unión de los
literales restantes.

Matemáticamente:

$$R=(C_i-\{l\})\cup(C_j-\{\neg l\})$$

Si el resolvente contiene un literal y su complementario,
se descarta porque constituye una tautología.

Si no existen literales complementarios, no se genera
ningún resolvente.


```python

# Cláusulas originales
ci = frozenset({"A", "B"})
cj = frozenset({"~B", "C"})

# Literal que se eliminará
l = "B"

# Eliminar los literales complementarios
parte1 = ci - {l}
parte2 = cj - {neg(l)}

# Unir los literales restantes
resolvente = parte1 | parte2

print("Cláusula 1:", ci)
print("Cláusula 2:", cj)

print("\nPrimera parte:", parte1)
print("Segunda parte:", parte2)

print("\nResolvente:", resolvente)
```


#### 4.5.1. Identificación de tautologías

Una cláusula es tautológica cuando contiene un literal
y su complementario.

Por ejemplo:

$$P\lor\neg P\lor Q$$

Esta cláusula siempre es verdadera, independientemente
de los valores de P y Q.

Por ello, no aporta restricciones adicionales a la
base de conocimiento y puede descartarse.

Se utilizará la función any() para comprobar si
algún literal tiene su complementario dentro de
la misma cláusula.


```python

def es_tautologia(clausula):
    """Determina si una cláusula contiene literales complementarios."""

    return any(neg(l) in clausula for l in clausula)


# Pruebas
clausulas_prueba = [
    frozenset({"P", "~P"}),
    frozenset({"P", "Q"}),
    frozenset({"P", "~P", "Q"}),
    frozenset()
]

for c in clausulas_prueba:
    print(c, "->", es_tautologia(c))
```

```python

def resolver(ci, cj):
    """Devuelve la lista de resolventes de dos cláusulas."""

    res = []

    # Recorrer los literales de la primera cláusula
    for l in sorted(ci):

        # Buscar el literal complementario
        if neg(l) in cj:

            # Eliminar los complementarios y unir
            nueva = (ci - {l}) | (cj - {neg(l)})

            # Descartar tautologías
            if not es_tautologia(nueva):

                # Evitar resolventes repetidos
                if nueva not in res:
                    res.append(nueva)

    return res
```


#### 4.5.2. Pruebas de la función resolver()

Estas pruebas permitirán verificar el comportamiento

### 4.6. Formato de presentación de cláusulas

Se implementará una función auxiliar que permita
mostrar las cláusulas utilizando símbolos de lógica
proposicional.

El símbolo "~" se sustituirá por "¬" y los literales
se separarán mediante el operador de disyunción "∨".

La cláusula vacía se representará mediante "□".

Esta función únicamente modifica la presentación
de los resultados, no el funcionamiento del algoritmo.


```python

def formato_clausula(clausula):
    """Convierte una cláusula a una expresión lógica legible."""

    if not clausula:
        return "□"

    literales = [
        l.replace("~", "¬")
        for l in sorted(clausula)
    ]

    return " ∨ ".join(literales)


# Pruebas de presentación
ejemplos = [
    frozenset({"P", "Q"}),
    frozenset({"~P", "Q"}),
    frozenset({"~B11", "P12", "P21"}),
    frozenset()
]

for c in ejemplos:
    print(formato_clausula(c))
```


### 4.7. Implementación del algoritmo PL-RESOLUTION

El algoritmo PL-RESOLUTION permite determinar si una
proposición es consecuencia lógica de una base de
conocimiento mediante resolución por refutación.

El procedimiento consiste en:

1. Incorporar las cláusulas de la base de conocimiento.
2. Agregar las cláusulas correspondientes a la meta negada.
3. Seleccionar pares de cláusulas.
4. Generar sus posibles resolventes.
5. Descartar tautologías y cláusulas repetidas.
6. Registrar los nuevos resolventes.
7. Repetir el procedimiento hasta encontrar la cláusula
   vacía o hasta que no se generen nuevas cláusulas.

Si se obtiene la cláusula vacía, el algoritmo devuelve True.

Si se agotan las posibilidades sin encontrar una
contradicción, devuelve False.


```python

def pl_resolution(clausulas, meta_negada):
    """True si se deriva la cláusula vacía; False en caso contrario."""

    # Incorporar la KB y la meta negada sin duplicados
    C = list(dict.fromkeys(
        frozenset(c) for c in clausulas + meta_negada
    ))

    # Asignar un identificador a cada cláusula
    ids = {c: i for i, c in enumerate(C, start=1)}

    ronda = 0

    print("CLÁUSULAS INICIALES")
    print("-" * 55)

    for c in C:
        print(f"C{ids[c]}: {formato_clausula(c)}")

    # Comprobar si ya existe la cláusula vacía
    if frozenset() in ids:
        print("La KB inicial contiene la cláusula vacía.")
        return True

    while True:

        ronda += 1
        nuevas = []

        print(f"\nRONDA {ronda}")
        print("-" * 55)

        # Seleccionar todos los pares de cláusulas
        for i, a in enumerate(C):

            for b in C[i + 1:]:

                # Obtener los resolventes del par
                for r in resolver(a, b):

                    # Registrar únicamente cláusulas nuevas
                    if r not in ids:

                        ids[r] = len(ids) + 1
                        nuevas.append(r)

                        print(
                            f"C{ids[r]}: {formato_clausula(r)} "
                            f"(C{ids[a]}, C{ids[b]})"
                        )

                        # Si se obtiene la cláusula vacía
                        if not r:
                            print("\nCláusula vacía encontrada.")
                            return True

        # Si no se generaron nuevas cláusulas
        if not nuevas:
            print("\nNo se generaron nuevas cláusulas.")
            return False

        # Incorporar los resolventes a la KB
        C.extend(nuevas)
```


#### 4.7.1. Prueba individual del algoritmo

Antes de ejecutar los ejercicios del manual se realizará
una prueba sencilla.

Base de conocimiento:

$$KB=\{P\}$$

Meta:

$$KB\models P$$

Al incorporar la meta negada, se obtiene:

$$P\land\neg P$$

La resolución de ambas cláusulas debe producir
la cláusula vacía.


```python

kb_prueba = [{"P"}]
meta_prueba = [{"~P"}]

resultado_prueba = pl_resolution(
    kb_prueba,
    meta_prueba
)

print("\nResultado:", resultado_prueba)
```

### 5.1. Registro de estadísticas del algoritmo

Para evaluar el desempeño del algoritmo PL-RESOLUTION
se registrarán las siguientes métricas:

- Cláusulas iniciales.
- Rondas de resolución.
- Pares de cláusulas evaluados.
- Resolventes nuevos.
- Cláusulas totales.
- Resultado de la demostración.

Estas métricas permitirán identificar la cantidad de
trabajo realizado por el algoritmo y compararla con
las demostraciones manuales.


```python

def pl_resolution_estadisticas(clausulas, meta_negada):
    """Ejecuta la resolución y registra estadísticas."""

    C = list(dict.fromkeys(
        frozenset(c) for c in clausulas + meta_negada
    ))

    conocidas = set(C)

    estadisticas = {
        "Cláusulas iniciales": len(C),
        "Rondas": 0,
        "Pares evaluados": 0,
        "Resolventes nuevos": 0,
        "Cláusulas totales": len(C),
        "Resultado": False
    }

    # Comprobar contradicción inicial
    if frozenset() in conocidas:
        estadisticas["Resultado"] = True
        return estadisticas

    while True:

        estadisticas["Rondas"] += 1

        nuevas = []

        for i, a in enumerate(C):

            for b in C[i + 1:]:

                estadisticas["Pares evaluados"] += 1

                for r in resolver(a, b):

                    if r not in conocidas:

                        conocidas.add(r)
                        nuevas.append(r)

                        estadisticas["Resolventes nuevos"] += 1
                        estadisticas["Cláusulas totales"] += 1

                        if not r:

                            estadisticas["Resultado"] = True

                            return estadisticas

        if not nuevas:
            return estadisticas

        C.extend(nuevas)
```


### 5.4. Estrategia de preferencia unitaria

La preferencia unitaria consiste en dar prioridad
a las resoluciones en las que al menos una de
las cláusulas contiene un único literal.

Estas cláusulas reciben el nombre de cláusulas
unitarias.

Por ejemplo:

$$P$$

$$\neg P\lor Q$$

Al aplicar resolución se obtiene:

$$Q$$

La estrategia permite obtener cláusulas más pequeñas
y reducir la cantidad de literales que deben analizarse.

En el ejercicio B se puede comenzar con la cláusula
unitaria P y obtener sucesivamente Q, R y S.

Finalmente, al resolver S con la meta negada ¬S,
se obtiene la cláusula vacía.

Esta estrategia podría reducir el trabajo necesario
para demostrar la meta.

Dar preferencia a las cláusulas unitarias no significa
que deban descartarse todas las demás resoluciones,
pues restringirse exclusivamente a ellas no garantiza
la completitud para cualquier conjunto de cláusulas.



### 5.5. Estrategia de conjunto de soporte

La estrategia de conjunto de soporte consiste en
dividir las cláusulas en dos grupos:

1. Las cláusulas de la base de conocimiento.
2. Las cláusulas del conjunto de soporte.

Inicialmente, el conjunto de soporte puede estar
formado por las cláusulas de la meta negada.

Cada resolución debe involucrar al menos una
cláusula perteneciente al conjunto de soporte.

Los resolventes obtenidos se incorporan también
a dicho conjunto.

De esta manera, la búsqueda se orienta hacia
las cláusulas relacionadas con la meta.

En el ejercicio B, el conjunto de soporte comienza
con la cláusula ¬S.

Al aplicar resolución se obtiene sucesivamente:

$$\neg S,\quad\neg R,\quad\neg Q,\quad\neg P$$

Finalmente, al resolver ¬P con P se obtiene
la cláusula vacía.

Esta estrategia evita resolver entre sí cláusulas
que no pertenecen al conjunto de soporte.

La estrategia conserva la completitud por refutación
cuando el conjunto de cláusulas fuera del soporte
es satisfacible.



### 5.6. Comparación de estrategias

| Estrategia | Funcionamiento | Ventaja | Limitación |
|---|---|---|---|
| Resolución exhaustiva | Examina todos los pares de cláusulas. | Búsqueda sistemática y completa. | Puede generar muchas cláusulas innecesarias. |
| Preferencia unitaria | Prioriza cláusulas de un solo literal. | Favorece la obtención de cláusulas pequeñas. | No siempre existen cláusulas unitarias disponibles. |
| Conjunto de soporte | Exige que al menos una cláusula pertenezca al conjunto de soporte. | Orienta la búsqueda hacia la meta. | Requiere seleccionar adecuadamente el conjunto de soporte. |

Para el ejercicio B, ambas estrategias permiten
identificar una demostración de cuatro resoluciones.

Se considera especialmente conveniente la estrategia
de conjunto de soporte, porque dirige el razonamiento
desde la meta negada hacia las cláusulas relevantes.

Las estrategias pueden combinarse para mejorar
la eficiencia del proceso de resolución.



### 5.7. Implementación del conjunto de soporte

Se implementará una versión alternativa del algoritmo
de resolución que comienza con la meta negada.

Los nuevos resolventes se incorporarán al conjunto
de soporte y se utilizarán para continuar la búsqueda.

Se contabilizarán los pares evaluados y los
resolventes nuevos para compararlos con el
algoritmo original.

Para los ejercicios de esta práctica, las bases de
conocimiento originales son satisfacibles.


```python

def pl_resolution_soporte(clausulas, meta_negada):
    """Resolución mediante conjunto de soporte."""

    C = list(dict.fromkeys(
        frozenset(c) for c in clausulas
    ))

    soporte = list(dict.fromkeys(
        frozenset(c) for c in meta_negada
    ))

    conocidas = set(C) | set(soporte)

    pares = 0
    resolventes = 0

    if frozenset() in conocidas:
        return True, pares, resolventes

    while soporte:

        # Extraer una cláusula del conjunto de soporte
        a = soporte.pop(0)

        for b in C:

            pares += 1

            for r in resolver(a, b):

                if r not in conocidas:

                    conocidas.add(r)
                    soporte.append(r)

                    resolventes += 1

                    print(
                        f"({formato_clausula(a)}) + "
                        f"({formato_clausula(b)}) "
                        f"-> {formato_clausula(r)}"
                    )

                    if not r:
                        return True, pares, resolventes

        # Incorporar la cláusula procesada
        if a not in C:
            C.append(a)

    return False, pares, resolventes
```

```python

print("EJERCICIO B: CONJUNTO DE SOPORTE\n")

resultado_soporte, pares_soporte, resolventes_soporte = (
    pl_resolution_soporte(kb_B, meta_B)
)

print("\nResultado:", resultado_soporte)
print("Pares evaluados:", pares_soporte)
print("Resolventes nuevos:", resolventes_soporte)
```

### Pregunta 1. ¿Por qué en (C) R no es consecuencia? Da un modelo de la KB donde R sea falsa.

R no es consecuencia lógica de la base de conocimiento
porque existe al menos una asignación de valores de
verdad que satisface todas las proposiciones de la KB
y en la cual R es falsa.

La base de conocimiento es:

$$KB=\{P\lor Q,\ P\Rightarrow R\}$$

Consideremos el siguiente modelo:

| Proposición | Valor de verdad |
|---|---|
| P | Falso |
| Q | Verdadero |
| R | Falso |

Evaluamos las expresiones:

$$P\lor Q=F\lor V=V$$

$$P\Rightarrow R=F\Rightarrow F=V$$

Ambas proposiciones de la KB son verdaderas,
aunque R es falsa.

Por lo tanto:

$$\boxed{KB\not\models R}$$

Este modelo constituye un contraejemplo que demuestra
que R no es consecuencia lógica de la base de
conocimiento.


```python

# Modelo propuesto
P = False
Q = True
R = False

# Evaluar las proposiciones de la KB
proposicion_1 = P or Q
proposicion_2 = not P or R

# Evaluar la base de conocimiento
kb_valida = proposicion_1 and proposicion_2

print("MODELO PROPUESTO")
print("-" * 40)

print("P:", P)
print("Q:", Q)
print("R:", R)

print("\nEVALUACIÓN")
print("-" * 40)

print("P ∨ Q:", proposicion_1)
print("P ⇒ R:", proposicion_2)
print("KB:", kb_valida)

print("\n¿Es un contraejemplo?", kb_valida and not R)
```

### Pregunta 3. La resolución es completa para refutación. ¿Qué significa eso en la práctica?

La completitud para refutación significa que, si
un conjunto de cláusulas es insatisfacible, existe
una secuencia de aplicaciones de la regla de
resolución que permite obtener la cláusula vacía.

Para demostrar una consecuencia lógica:

$$KB\models\alpha$$

Se incorpora la negación de la meta:

$$KB\land\neg\alpha$$

Si esta expresión es insatisfacible, un procedimiento
de resolución completo podrá encontrar una
contradicción:

$$\square$$

En la práctica, esto significa que, si una proposición
es realmente consecuencia lógica de una base de
conocimiento proposicional, el algoritmo exhaustivo
de resolución podrá demostrarla.

Si el algoritmo alcanza la saturación sin obtener
la cláusula vacía, la meta no es consecuencia
lógica de la KB.

La completitud no garantiza que la demostración
sea rápida o que utilice la menor cantidad posible
de resoluciones.

del conocimiento disponible. Su eficiencia depende
tanto de la representación del conocimiento como
de las estrategias utilizadas para seleccionar
las cláusulas durante la inferencia.
